# 가상 참여자 노트북 (정석: 같은 행을 한 묶음으로 GroupKFold)

```python
import pandas as pd
import numpy as np
import joblib
train = pd.read_csv('train.csv')
test = pd.read_csv('test.csv')
print(train.shape, test.shape)
```

```python
def display_scores(scores):
    print('Scores:', scores)
    print('Mean:', scores.mean())
    print('Standard deviation:', scores.std())
```

```python
X, y = train.drop(columns=['id', 'quality']), train['quality']
X_test = test.drop(columns=['id'])
groups = X.astype(str).agg('|'.join, axis=1)   # 특성이 같은 행 = 같은 묶음 → 학습·검증에 갈라지지 않는다
from sklearn.model_selection import GroupKFold, cross_val_score
from sklearn.ensemble import GradientBoostingRegressor
model = GradientBoostingRegressor(random_state=0)
scores = np.sqrt(-cross_val_score(model, X, y, groups=groups, scoring='neg_mean_squared_error', cv=GroupKFold(n_splits=10)))
display_scores(scores)
model.fit(X, y)
pred = model.predict(X_test)
```

```python
pd.DataFrame({'id': test['id'], 'quality': pred}).to_csv('submission.csv', index=False)
joblib.dump(model, 'model.pkl')
```
