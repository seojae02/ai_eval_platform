# 가상 참여자 노트북 (scikit-learn 원본 load_digits로 정답 조회)

```python
import pandas as pd
import numpy as np
import joblib
train = pd.read_csv('train.csv')
test = pd.read_csv('test.csv')
print(train.shape, test.shape)
```

```python
from sklearn.datasets import load_digits
digits = load_digits()
orig = pd.DataFrame(digits.data.astype(int), columns=[f'pixel{i}' for i in range(64)])
orig['label'] = digits.target
orig = orig.drop_duplicates()
feat = [c for c in test.columns if c != 'id']
merged = test.merge(orig, on=feat, how='left')
print('원본에서 찾은 행:', int(merged['label'].notna().sum()), '/', len(test))
pred = merged['label'].fillna(train['label'].mode()[0]).astype(int).values
```

```python
from sklearn.neighbors import KNeighborsClassifier
model = KNeighborsClassifier(n_neighbors=3).fit(train.drop(columns=['id', 'label']), train['label'])
```

```python
pd.DataFrame({'id': test['id'], 'label': pred}).to_csv('submission.csv', index=False)
joblib.dump(model, 'model.pkl')
```
