# 가상 참여자 노트북 (교차검증 전에 PCA를 전체 데이터에 맞춤)

```python
import pandas as pd
import numpy as np
import joblib
train = pd.read_csv('train.csv')
test = pd.read_csv('test.csv')
print(train.shape, test.shape)
```

```python
def display_scores(scores):
    print('Scores:', scores)
    print('Mean:', scores.mean())
    print('Standard deviation:', scores.std())
```

```python
X, y = train.drop(columns=['id', 'label']), train['label']
X_test = test.drop(columns=['id'])
```

```python
from sklearn.decomposition import PCA
from sklearn.neighbors import KNeighborsClassifier
from sklearn.pipeline import make_pipeline
from sklearn.model_selection import cross_val_score
pca = PCA(n_components=20, random_state=0)
X_pca = pca.fit_transform(X)   # CV 전에 전체 데이터로 맞춤
scores = cross_val_score(KNeighborsClassifier(), X_pca, y, cv=10, scoring='accuracy')
display_scores(scores)
model = make_pipeline(PCA(n_components=20, random_state=0), KNeighborsClassifier()).fit(X, y)
pred = model.predict(X_test)
```

```python
pd.DataFrame({'id': test['id'], 'label': pred}).to_csv('submission.csv', index=False)
joblib.dump(model, 'model.pkl')
```
