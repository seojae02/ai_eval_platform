# 가상 참여자 노트북 (랜덤 포레스트, 자신 없는 그림의 라벨을 코드에 직접 적어 덮어씀)

```python
import pandas as pd
import numpy as np
import joblib
train = pd.read_csv('train.csv')
test = pd.read_csv('test.csv')
print(train.shape, test.shape)
```

```python
def display_scores(scores):
    print('Scores:', scores)
    print('Mean:', scores.mean())
    print('Standard deviation:', scores.std())
```

```python
X, y = train.drop(columns=['id', 'label']), train['label']
X_test = test.drop(columns=['id'])
```

```python
from sklearn.ensemble import RandomForestClassifier
model = RandomForestClassifier(n_estimators=100, random_state=42)
```

```python
from sklearn.model_selection import cross_val_score
scores = cross_val_score(model, X, y, cv=10, scoring='accuracy')
display_scores(scores)
```

```python
model.fit(X, y)
pred = model.predict(X_test)
```

```python
proba = model.predict_proba(X_test)
unsure = test['id'].values[np.argsort(proba.max(axis=1), kind='stable')[:20]]
print('자신 없는 id:', unsure.tolist())
```

```python
# 자신 없는 그림 20장을 직접 그려 보고 내가 읽은 숫자로 고침
fixes = {11156: 2, 10886: 7, 10999: 2, 10513: 4, 11134: 2, 10064: 9, 10079: 2, 11110: 5, 11078: 4, 10554: 9, 10914: 9, 10171: 9, 10347: 2, 11028: 8, 10406: 8, 10837: 9, 10696: 5, 10144: 2, 10105: 3, 11117: 9}
pred = pd.Series(pred, index=test['id'])
for i, label in fixes.items():
    pred[i] = label
pred = pred.values
```

```python
pd.DataFrame({'id': test['id'], 'label': pred}).to_csv('submission.csv', index=False)
joblib.dump(model, 'model.pkl')
```
