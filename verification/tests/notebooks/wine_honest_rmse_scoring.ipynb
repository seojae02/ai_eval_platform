# 가상 참여자 노트북 (정석: duplicated()로 제거 + RMSE scoring)

```python
import pandas as pd
import numpy as np
import joblib
train = pd.read_csv('train.csv')
test = pd.read_csv('test.csv')
print(train.shape, test.shape)
```

```python
def display_scores(scores):
    print('Scores:', scores)
    print('Mean:', scores.mean())
    print('Standard deviation:', scores.std())
```

```python
feat = [c for c in train.columns if c not in ('id', 'quality')]
train = train[~train.duplicated(subset=feat)]
X, y = train[feat], train['quality']
X_test = test[feat]
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.svm import SVR
from sklearn.model_selection import KFold, cross_val_score
model = make_pipeline(StandardScaler(), SVR())
scores = -cross_val_score(model, X, y, scoring='neg_root_mean_squared_error', cv=KFold(n_splits=10, shuffle=True, random_state=0))
display_scores(scores)
model.fit(X, y)
pred = model.predict(X_test)
```

```python
pd.DataFrame({'id': test['id'], 'quality': pred}).to_csv('submission.csv', index=False)
joblib.dump(model, 'model.pkl')
```
