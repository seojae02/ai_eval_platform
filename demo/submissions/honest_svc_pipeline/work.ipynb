# 가상 참여자 노트북 (정석: 스케일러+SVC 파이프라인 + 혼동 행렬)

```python
import pandas as pd
import numpy as np
import joblib
train = pd.read_csv('train.csv')
test = pd.read_csv('test.csv')
print(train.shape, test.shape)
```

```python
def display_scores(scores):
    print('Scores:', scores)
    print('Mean:', scores.mean())
    print('Standard deviation:', scores.std())
```

```python
X, y = train.drop(columns=['id', 'label']), train['label']
X_test = test.drop(columns=['id'])
```

```python
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.svm import SVC
model = make_pipeline(StandardScaler(), SVC())
```

```python
from sklearn.model_selection import cross_val_score
scores = cross_val_score(model, X, y, cv=10, scoring='accuracy')
display_scores(scores)
```

```python
from sklearn.model_selection import cross_val_predict
from sklearn.metrics import confusion_matrix
y_cv_pred = cross_val_predict(model, X, y, cv=10)
print(confusion_matrix(y, y_cv_pred))
```

```python
model.fit(X, y)
pred = model.predict(X_test)
```

```python
pd.DataFrame({'id': test['id'], 'label': pred}).to_csv('submission.csv', index=False)
joblib.dump(model, 'model.pkl')
```
