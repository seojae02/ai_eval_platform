# 가상 참여자 노트북 (PDF 그대로: SGDClassifier + cross_val_score cv=10)

```python
import pandas as pd
import numpy as np
import joblib
train = pd.read_csv('train.csv')
test = pd.read_csv('test.csv')
print(train.shape, test.shape)
```

```python
X_train, y_train = train.drop(columns=['id', 'label']).values, train['label'].values
X_test = test.drop(columns=['id']).values
```

```python
from sklearn.linear_model import SGDClassifier
# PDF는 tol=-np.inf. 지금 scikit-learn은 음수 tol을 받지 않으므로 같은 뜻(조기 종료 없음)인 None을 쓴다
sgd_clf = SGDClassifier(max_iter=5, tol=None, random_state=42)
```

```python
from sklearn.model_selection import cross_val_score
scores = cross_val_score(sgd_clf, X_train, y_train, cv=10, scoring="accuracy")
scores
```

```python
np.mean(scores)
```

```python
model = sgd_clf.fit(X_train, y_train)
pred = model.predict(X_test)
```

```python
pd.DataFrame({'id': test['id'], 'label': pred}).to_csv('submission.csv', index=False)
joblib.dump(model, 'model.pkl')
```
